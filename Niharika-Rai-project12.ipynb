```python
class Flight:
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay): 
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay 
    
    def get_arrdelay(self): 
        return self.arrival_delay 
    
class ScheduledFlight(Flight): 
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay, CRSDepTime, CRSArrTime):
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay
        self.CRSDepTime = CRSDepTime
        self.CRSArrTime = CRSArrTime 
        
    def is_ontime(self):
        if self.CRSDepTime==self.departure_time:
            return True 
        else: 
            return False 
        if self.CRSArrTime==self.arrival_time:
            return True
        else: 
            return False 
```

```python
class Flight:
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay): 
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay 
    
    def get_arrdelay(self): 
        return self.arrival_delay 
    
class ScheduledFlight(Flight): 
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay, CRSDepTime, CRSArrTime):
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay
        self.CRSDepTime = CRSDepTime
        self.CRSArrTime = CRSArrTime 
        
    def is_ontime(self):
        if self.CRSDepTime==self.departure_time:
            return True 
        else: 
            return False 
        if self.CRSArrTime==self.arrival_time:
            return True
        else: 
            return False 
        
    def is_delayed(self): 
        if self.CRSDepTime==self.departure_time:
            return False
        else: 
            return True 
        if self.CRSArrTime==self.arrival_time: 
            return False
        else:
            return True 
```

```python
longlistofflights = [] 
for index, row in myDF.iterrows(): 
    myflight = ScheduledFlight(
        flight_number = row['Flight_Number_Reporting_Airline'],
        origin_airport_ID = row['Origin'],
        destination_airport_ID = row['Dest'],
        departure_time = row['DepTime'],
        arrival_time = row['ArrTime'],
        departure_delay = row['DepDelay'], 
        arrival_delay = row['ArrDelay'], 
        CRSDepTime = row['CRSDepTime'], 
        CRSArrTime = row['CRSArrTime']
    
    )
    longlistofflights.append(myflight) 
```

```python
delayed_count = {} 
for myflight in longlistofflights: 
    if myflight.destination_airport_ID not in delayed_count:
        delayed_count[myflight.destination_airport_ID] = [] 
    delayed_count[myflight.destination_airport_ID].append(myflight.is_delayed()) 
```

```python
class Flight:
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay): 
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay 
    
    def get_arrdelay(self): 
        return self.arrival_delay 
    
class CancelledFlight(Flight): 
    def __init__(self, flight_number, origin_airport_ID, destination_airport_ID, departure_time, arrival_time, departure_delay, arrival_delay, CRSDepTime, CRSArrTime):
        self.flight_number = flight_number 
        self.origin_airport_ID = origin_airport_ID
        self.destination_airport_ID = destination_airport_ID
        self.departure_time = departure_time
        self.arrival_time = arrival_time
        self.departure_delay = departure_delay
        self.arrival_delay = arrival_delay
        self.CRSDepTime = CRSDepTime
        self.CRSArrTime = CRSArrTime  
        
    def get_cancelled_flight(self):
        if self.CRSDepTime==0:
            return True
        else: 
            return False
        if self.CRSArrTime==0:
            return True
        else:
            return False 
```

```python
longlistofflights = [] 
for index, row in myDF.iterrows(): 
    myflight = CancelledFlight(
        flight_number = row['Flight_Number_Reporting_Airline'],
        origin_airport_ID = row['Origin'],
        destination_airport_ID = row['Dest'],
        departure_time = row['DepTime'],
        arrival_time = row['ArrTime'],
        departure_delay = row['DepDelay'], 
        arrival_delay = row['ArrDelay'], 
        CRSDepTime = row['CRSDepTime'], 
        CRSArrTime = row['CRSArrTime']
    
    )
    longlistofflights.append(myflight) 
```

```python
cancelled_flight_count = {} 
for myflight in longlistofflights: 
    if myflight.destination_airport_ID not in cancelled_flight_count:
        cancelled_flight_count[myflight.destination_airport_ID] = [] 
    cancelled_flight_count[myflight.destination_airport_ID].append(myflight.get_cancelled_flight()) 
```

```python
total_cancelled_flight_count = {myairport: sum(cancelled_flight_count) for myairport, cancelled_flight_count in cancelled_flight_count.items()} 
```
