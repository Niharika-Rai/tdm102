## Question 5

```python
import pandas as pd
from pathlib import Path
myfiles=[]
for year in range (1880, 1884):
    file= Path(f'/anvil/projects/tdm/data/noaa/{year}.csv')
    myfiles.append(file) 
    
count = 0
for file in myfiles:
    for myDF in pd.read_csv(file,names=["id","date","element_code","value","mflag","qflag","sflag","obstime"],chunksize =10000):
        count += len(myDF[myDF['element_code'] == 'SNOW'])

print(count) 
```

```python
count = 0
for file in myfiles:
    for myDF in pd.read_csv(file,names=["id","date","element_code","value","mflag","qflag","sflag","obstime"],chunksize =10000):
        for index, row in myDF.iterrows():
            if row['element_code'] == 'SNOW':
                count += 1

print(count) 
```

For part a, I got all the codes from the question. Both versions of the program lead to the same answer. This answer is found by adding: 33700+36451+40648+53417 = 164216. This is because the question asked to find how many times 'SNOW' occured in each of the data frames and hence, the answer was the sum of the values in each of the 4 data frames where 'SNOW' occurred. According to this website: https://www.geeksforgeeks.org/how-to-load-a-massive-file-as-small-chunks-in-pandas/, chunksize is basicially the number of rows that are read at 1 given time by the file in pandas. So, if the chunksize in this case is 10000 and the number of rows in the 1880.csv file is 370779, then the number of rows read at once are in 370779/10000 = 37 chunks. The remaining 0779 form the 38th chunk. However, another important point I found out from the same website was that chunksize does not affect the columns in anyway. So, the 8 columns that we have in each of the data frames from 1880.csv to 1883.csv stay as they are.
