## Question 5

```python
import time
start_time = time.time()

import numpy as np 

my_departure_array = myDF['DepDelay'].to_numpy()
my_arrival_array = myDF['ArrDelay'].to_numpy()
my_distance_array = myDF['Distance'].to_numpy()

my_filtered_distances = my_distance_array[(my_departure_array > 60) | (my_arrival_array > 60)]
my_average_distance_with_numpy = np.mean(my_filtered_distances)

print(f"The average distance for those delayed flights is {my_average_distance_with_numpy}")

end_time = time.time()
print(f"Used time is {end_time - start_time}")
```
