```python
import matplotlib.pyplot as plt
my_temperature = 90.0 
myresults = myDF[myDF['temperature'] == my_temperature].groupby(['year', 'month'])['humidity'].mean() 
myresults.unstack(0).plot(kind = 'bar') 
plt.title(f'Average Humidity For The Given Temperature And All Month-And-Year Pairs')
plt.xlabel('Month')
plt.ylabel('Average Humidity')
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.legend(title = 'Year')
plt.grid(True)
plt.show() 
```

```python
def bar_plot (my_temperature):
    import matplotlib.pyplot as plt
    myresults = myDF[myDF['temperature'] == my_temperature].groupby(['year', 'month'])['humidity'].mean() 
    final_result = myresults.unstack(0).plot(kind = 'bar') 
    plt.title(f'Average Humidity For The Given Temperature And All Month-And-Year Pairs')
    plt.xlabel('Month')
    plt.ylabel('Average Humidity')
    plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
    plt.legend(title = 'Year')
    plt.grid(True)
    plt.show()
    return final_result 
```
